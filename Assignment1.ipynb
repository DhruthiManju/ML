```python
#Question 3 Element-wise Arithmetic and Aggregations
#step 1:Create the arrays
prices=np.array([120,250,80,450,300])
quantities=np.array([4,2,10,1,3])
```

```python
#Step 2: Calculate individual product revenue
#Revenue=Price*Quantity
#We didn't write a loop to multiply individual elements.
#NumPy automatically performs the operation across the arrays. This is called vectorization.
revenue=prices*quantities
print("Revenue:",revenue)
```

```python
#Step 3: Calculate combined revenue
#sum=480+500+800+450+900=3130
total_revenue=np.sum(revenue)
print("Total Revenuee:",total_revenue)
```

```python
#Step 4: Calculate average revenue
#avg=3130//5=626
avg_revenue=np.mean(revenue)
print("Average Revenue:",avg_revenue)
```

```python
#Step 5: Find the highest-earning product
highest_index=np.argmax(revenue)
print("highest Earning Product Index:",highest_index)
```

```python
#Question 4: Boolean Masking and Data Cleaning
#Step 1: Create the rainfall array
rainfall = np.array([
    12.5, -1.0, 0.0, 18.2, -99.0,
    5.4, 22.1, 0.0, -1.0, 14.8
])
```

```python
#Step 2: Extract valid rainfall values[Boolean Condition]
valid_condition=rainfall>=0
print(valid_condition)
```

```python
#Boolean Masking
valid_rainfall=rainfall[rainfall>=0]
print("Valid Rainfall:",valid_rainfall)
```

```python
#Step 3: Replace negative values with zero
#We can use Boolean indexing to modify the original array.
rainfall[rainfall<0]=0.0
print("Cleaned Rainfall:",rainfall)
```

```python
#Step 4: Count heavy rainfall days
#Heavy rainfall means strictly greater than 10 mm.
heavy_rainfall=rainfall[rainfall>10]
count=np.count_nonzero(rainfall>10)
print("Heavy Rainfall Values:",heavy_rainfall)
print("Number of Heavy Rainfall Days:",count)
```

```python
#Question 5: Conditional Array Operations with np.where()
#Step 1: Generate random marks
import numpy as np
np.random.seed(10)
scores=np.random.randint(20,101,12)
print("Original Scores:",scores)
```

```python
#Step 2: Create Pass and Fail status
status=np.where(scores>=40,"Pass","Fail")
print("Student Status:",status)
```

```python
#Step 3: Add grace marks
updated_scores=np.where(scores>40,scores+5,scores)
print("Updateed Scores:",updated_scores)
#Notice that the updated scores are stored in a new array. The original scores array remains unchanged.
```

```python
#Question 6: Row-wise and Column-wise Aggregations
#Step 1: Generate the marks matrix
marks=np.random.randint(50,101,(6,4))
print("marks Matrix:")
print(marks)
```

```python
#Question 7: Feature Scaling via Broadcasting
#Step 1: Create the dataset
houses = np.array([
    [1200, 2, 55],
    [1800, 3, 85],
    [950, 1, 40],
    [2400, 4, 120],
    [1500, 3, 70]
])
print(houses)
```

```python
#Step 2: Find the minimum value of each column
minimum=np.min(houses,axis=0)
print("Minimum Values:",minimum)
```

```python
#Step 3: Find the maximum value of each column
maximum=np.max(houses,axis=0)
print("Maximum Values:",maximum)
```

```python
#Step 4: Apply Min-Max Normalization
normalized=(houses-minimum)/(maximum-minimum)
print("normalized Data:")
print(np.round(normalized,3))
```

How can NumPy subtract an array of shape (3,) from an array of shape (5, 3)?
bcs houses.shape=(5,3) and minimum.shape=(3,)
This is possible because of broadcasting.

Broadcasting allows NumPy to perform operations between arrays with compatible shapes.
Conceptually, NumPy uses the minimum values for every row:
[950, 1, 40]
[950, 1, 40]
[950, 1, 40]
[950, 1, 40]
[950, 1, 40]

```python
#Question 8: Combining and Stacking Arrays
#Step 1: Create morning sales
morning_sales = np.array([
    [15, 20],
    [10, 12],
    [25, 30]
])
```

```python
#Step 2: Create evening sales
evening_sales = np.array([
    [18, 22],
    [14, 15],
    [20, 28]
])
```

```python
#Step 3: Combine arrays horizontally
full_day_sales=np.hstack((morning_sales,evening_sales))
print("Full Day Sales:")
print(full_day_sales)
```

```python
#Step 4: Add the fourth branch
fourth_branch = np.array([12, 16, 15, 19])
final_sales=np.vstack((full_day_sales,fourth_branch))
print("Final Sales Matrix:")
print(final_sales)
```

```python
#Step 5: Calculate total units sold in each column
column_totals=np.sum(final_sales,axis=0)
print("Total Units Sold in Each Column:",column_totals)
```

```python
#Question 9: Unique Elements and Frequency Distribution
#Step 1: Create the ratings array
ratings = np.array([
    5, 4, 3, 5, 2, 5, 4, 1, 3, 5,
    4, 4, 2, 5, 3, 4, 5, 1, 4, 3
])

print("Ratings:", ratings)
```

```python
#Step 2: Find unique ratings
unique_ratings=np.unique(ratings)
print("Unique Ratings:",unique_ratings)
```

```python
#Step 3: Find the frequency of each rating
unique_ratings,counts=np.unique(ratings,return_counts=True)
print("Unique Ratings:",unique_ratings)
print("Frequency Counts:",counts)
```

```python
#Step 4: Calculate percentage share
#percentage=count/total*100
percentage=(counts/ratings.size)*100
print("Percentage:",percentage)
```

```python
#Step 5: Print a clean summary
summary=np.column_stack((unique_ratings,counts,percentage))
print("Rating Summary:")
print(summary)
```

```python
#Question 10: Indirect Sorting and Top-K Selection
#Step 1: Create employee arrays
emp_ids = np.array([
    "E101", "E102", "E103",
    "E104", "E105", "E106"
])

scores = np.array([78, 92, 65, 88, 95, 71])
```

```python
#Step 2: Understand np.argsort()
sorted_indices=np.argsort(scores)
print("Sorted Indices:",sorted_indices)
```

```python
#Step 3: Sort in descending order
descending_indices=np.argsort(scores)[::-1]
print("Descending Indices:",descending_indices)
```

```python
#Step 4: Reorder employee IDs and scores
sorted_emp_ids=emp_ids[descending_indices]
sorted_scores=scores[descending_indices]
print("Sorted Employee IDs:",sorted_emp_ids)
print("Sorted Scores:",sorted_scores)
```

```python
#Step 5: Find the Top 3 employees
top_3_ids=sorted_emp_ids[:3]
top_3_scores=sorted_scores[:3]
print("Top 3 Employees:")
print(top_3_ids)
print(top_3_scores)
```
