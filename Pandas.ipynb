```python
#Understanding rows, columns, index and shape
df = pd.DataFrame({
    "Name": ["Rahul", "Priya", "John", "Anu"],
    "Age": [22, 23, 21, 22],
    "Marks": [80, 90, 75, 85]
})
print(df.shape)
print(df.index)
print(df.columns)
print(df.dtypes)
print(df.size) #4*3=12
print(df.ndim)
```

```python
#Viewing and Understanding Data
#1.Head():Displays first 5 rows among 100000 records of dataset
print(df.head(2))
print(df.tail(2))
print(df.info())
print(df.describe())
print(df.sample(3))
```

```python
#6.Selecting Columns and Rows
df = pd.DataFrame({
    "Name": ["Rahul", "Priya", "John", "Anu"],
    "Age": [22, 23, 21, 22],
    "Marks": [80, 90, 75, 85]
})
#A. Selecting a single column
print(df["Name"])
#B. Selecting multiple columns
print(df[["Name","Marks"]])
#C. Selecting rows using iloc----iloc means integer-location based indexing.
# Syntax  --> df.iloc[row_postion,column_position]
print(df.iloc[0])
print(df.iloc[0:2])
#D. Selecting rows using loc-----loc means label-based indexing.
print(df.loc[0])
print(df.loc[0:2, "Name"])
#7. Filtering Data
result=df[df["Marks"]>80]
print(result)
```

```python
#8.Adding, Updating and Deleting Columns
#A. Adding a new column
df["Department"]=[
    "IT","HR","IT","finance"]
print(df)
#B. Adding a calculated column
df["Bonus"]=df["Marks"]+5
print(df)
#C. Updating values
df.loc[0,"Marks"]=88
print(df)
#D. Deleting a column
# df.drop(columns=["Bonus"], inplace=True)
new_df=df.drop(columns=["Bonus"])
print(new_df)
print(df)
```

```python
#9. Handling Missing Values (NaN)
df = pd.DataFrame({
    "Name": ["Rahul", "Priya", "John", "Anu"],
    "Age": [22, None, 25, 23],
    "Salary": [30000, 40000, None, 35000]
})

print(df)
```

```python
#A. isnull()
print(df.isnull())
#B. isnull().sum()
print(df.isnull().sum())
#C. dropna()-----Removes rows containing missing values.
new_df=df.dropna()
print(new_df)
#D. fillna()----Instead of deleting rows, we can fill missing values.
df["Age"]=df["Age"].fillna(df["Age"].mean())
df["Salary"]=df["Salary"].fillna(df["Salary"].mean())
```

```python
#10. Sorting Data
#Ascending order
df = pd.DataFrame({
    "Name": ["Rahul", "Priya", "John", "Anu"],
    "Marks": [80, 90, 75, 85]
})

print(df.sort_values(by="Marks"))
#Descending Order
print(df.sort_values(by="Marks",ascending=False))
```

```python
#11.Group By----Used for Data Analysis 
df = pd.DataFrame({
    "Name": ["Rahul", "Priya", "John", "Anu", "Kiran"],
    "Department": ["IT", "HR", "IT", "HR", "IT"],
    "Salary": [30000, 40000, 50000, 35000, 45000]
})
# result=df.groupby("Department")["Salary"].mean()
# result=df.groupby("Department")["Salary"].sum()
# result=df.groupby("Department")["Salary"].max()
# result=df.groupby("Department")["Salary"].min()
# result=df.groupby("Department")["Salary"].count()
# print(result)
#OR
result=df.groupby("Department")["Salary"].agg(["mean","sum","min","max","count"])
print(result)
```

```python
#12. Applying Functions – apply(), map() and applymap()
#A. apply()---Used to apply a function a Series or DataFrame axis
df=pd.DataFrame({"Marks":[40,60,80,90]})
def check_result(marks):
    if marks>=50:
        return "Pass"
    else:
        return "Fail"
df["Result"]=df["Marks"].apply(check_result)
print(df)
```

```python
#B. map()-----Used to map values in a Series using a dictionary or function.
df = pd.DataFrame({"Department": ["IT", "HR", "IT", "Finance"]})
department_names = {
    "IT": "Information Technology",
    "HR": "Human Resources",
    "Finance": "Finance Department"
}
df["Full Department"]=df["Department"].map(department_names)
print(df)

#For DataFrame element-wise transformations, newer Pandas versions also provide DataFrame.map(). Older tutorials may call this applymap(), which is deprecated in modern Pandas.
```

```python
#14. Combining Multiple DataFrames
#A.concat()
df1 = pd.DataFrame({
    "Name": ["Rahul", "Priya"],
    "Age": [22, 23]
})

df2 = pd.DataFrame({
    "Name": ["John", "Anu"],
    "Age": [25, 24]
})
result=pd.concat([df1,df2],ignore_index=True)  # ignore_index=True is used to avoid repeated index during joining and give continous index
print(result)
```

```python
#B. merge()
df1 = pd.DataFrame({
    "Emp_ID": [101, 102, 103],
    "Name": ["Rahul", "Priya", "John"]
})

df2 = pd.DataFrame({
    "Emp_ID": [101, 102, 103],
    "Salary": [30000, 40000, 50000]
})

result = pd.merge(df1,df2,on="Emp_ID")
result1 = pd.merge(df1,df2,on="Emp_ID",how="inner")
result2= pd.merge(df1,df2,on="Emp_ID",how="left")
result3= pd.merge(df1,df2,on="Emp_ID",how="right")
result4= pd.merge(df1,df2,on="Emp_ID",how="outer")

print(result)
print(result1)
print(result2)
print(result3)
print(result4)
```

```python
#C. join()
df1 = pd.DataFrame(
    {"Name": ["Rahul", "Priya"]},
    index=[101, 102]
)

df2 = pd.DataFrame(
    {"Salary": [30000, 40000]},
    index=[101, 102]
)

result = df1.join(df2)

print(result)
```

```python
#15. Removing Duplicate Values
#A.Duplicated()
df = pd.DataFrame({
    "Name": ["Rahul", "Priya", "Rahul", "John"],
    "Age": [22, 23, 22, 25]
})

print(df.duplicated())
```

```python
#B. drop_duplicates()
df=df.drop_duplicates()
print(df)
```

```python
#C. Resetting the index  -----After deleting rows, you might want to reset the index.
df=df.reset_index(drop=True)
print(df)
```

```python
#16. Working with Text Data
df = pd.DataFrame({
    "Name": ["rahul", "PRIYA", "john", "anu"]
})
#A.Convert to uppercase
df["Name"]=df["Name"].str.upper()
print(df)
#B. Convert to lowercase
df["Name"]=df["Name"].str.lower()
print(df)
#C. Convert to title case
df["Name"]=df["Name"].str.title()
print(df)
```

```python
#D. Check whether text contains a particular word
df = pd.DataFrame({
    "Email": [
        "rahul@gmail.com",
        "priya@yahoo.com",
        "john@gmail.com"
    ]
})

result = df[df["Email"].str.contains("gmail", case=False)]

print(result)
```

```python
#17. Working with Date and Time
df = pd.DataFrame({
    "Name": ["Rahul", "Priya", "John"],
    "Joining_Date": [
        "2025-01-10",
        "2025-03-15",
        "2025-06-20"
    ]
})
#A. to_datetime()
df["Joining_Date"] = pd.to_datetime(df["Joining_Date"])

print(df.dtypes)
```

```python
#B. Extract year
df["Year"] = df["Joining_Date"].dt.year
print(df["Year"])
```

```python
#C. Extract month
df["Month"] = df["Joining_Date"].dt.month
print(df["Month"])
```

```python
#D. Extract day
df["Day"]=df["Joining_Date"].dt.day
print(df["Day"])
```

```python
#18. Pivot Tables ---A pivot table summarizes data by arranging it according to rows and columns.
df = pd.DataFrame({
    "Department": ["IT", "IT", "HR", "HR", "IT"],
    "Gender": ["Male", "Female", "Male", "Female", "Male"],
    "Sales": [100, 200, 150, 250, 300]
})
result=pd.pivot_table(df,values="Sales",index="Department",columns="Gender",aggfunc="mean")
print(result)
```

```python
#Data Analytics Advance Concepts
#A.value_counts()
df = pd.DataFrame({
    "Department": [
        "IT", "HR", "IT", "Finance", "IT", "HR"
    ]
})

print(df["Department"].value_counts())
```

```python
# D. corr()----Correlation measures the strength and direction of a linear relationship between numerical variables.
df = pd.DataFrame({
    "Study_Hours": [1, 2, 3, 4, 5],
    "Marks": [20, 40, 50, 70, 90]
})

print(df.corr())
```

```python
#E. rolling()----Rolling calculates values over a moving window.
sales = pd.Series([100, 200, 300, 400, 500])

print(sales.rolling(window=3).mean())
```
