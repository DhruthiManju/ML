```python
#Step 1&2: Check missing values
df.isnull().sum()
```

```python
# Step 3: Find the percentage of missing values
missing_count=df.isnull().sum()
missing_percentage=(df.isnull().sum()/len(df))*100
missing_data=pd.DataFrame({
    "Missing Count":missing_count,
    "Missing Percentage":missing_percentage
})
print(missing_data)
```

```python
# Handling Missing Values – Cleaning the Dataset
# Cleaning the Age column
# Calculate median age
print(df["Age"].median())
# Fill missing Age values
df["Age"]=df["Age"].fillna(df.groupby(["Pclass","Sex"])["Age"]
.transform("median"))
print(df["Age"])
print(df["Age"].isnull().sum())
```

```python
# Cleaning the Embarked column
# Step 1: Find the mode
print(df["Embarked"].mode())
```

```python
# Step 2: Fill missing values
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
```

```python
print(df["Embarked"].isnull().sum())
```

```python
 # Cleaning the Cabin column
df["Cabin"] = df["Cabin"].fillna("Unknown")
```

```python
print(df["Cabin"].isnull().sum())
```

```python
# Cabin numbers often start with a letter.so extract the letter
df["Deck"] = df["Cabin"].str[0]
print(df[["Cabin", "Deck"]].head(10))
```
